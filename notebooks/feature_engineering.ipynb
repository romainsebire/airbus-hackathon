```python
import matplotlib.pyplot as plt
import pandas as pd
```

```python
df_train = pd.read_csv("data/clean_train.csv")
base_cols_train = ["id", "wip", "investissement", "satisfaction"]
```

```python
demand_cols = [c for c in df_train.columns if c.startswith("demand")]

df_demand = df_train[base_cols_train + demand_cols]

df_demand.to_csv(
    "data/clean_train_demand_only.csv",
    index=False
)
```

```python
param_cols = [c for c in df_train.columns if c.startswith("param")]

df_param = df_train[base_cols_train + param_cols]

df_param.to_csv(
    "data/clean_train_param_only.csv",
    index=False
)
```

```python
df_test = pd.read_csv("data/clean_test.csv")

base_test_cols = ["id"]

demand_cols_test = [c for c in df_test.columns if c.startswith("demand")]
param_cols_test  = [c for c in df_test.columns if c.startswith("param")]

df_test_demand = df_test[base_test_cols + demand_cols_test]
df_test_param  = df_test[base_test_cols + param_cols_test]

df_test_demand.to_csv("data/clean_test_demand_only.csv", index=False)
df_test_param.to_csv("data/clean_test_param_only.csv", index=False)
```
